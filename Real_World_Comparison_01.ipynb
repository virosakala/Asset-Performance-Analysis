```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# 1. Defining the assets (Gold ETF and Bitcoin)
assets = ['GLD', 'BTC-USD']

# 2. Downloading 1 year of historical daily data
raw_data = yf.download(assets, period='1y')['Close']

# 3. Normalizing the data to show percentage growth (Price / Initial Price)
# This allows us to compare a $2000 asset and a $60,000 asset on the same scale
normalized_data = (raw_data / raw_data.iloc[0]) * 100

# 4. Visualizing the divergence
plt.style.use('dark_background')
normalized_data.plot(figsize=(12, 6), color=['gold', 'cyan'])
plt.title('Asset Performance Comparison (Last 12 Months)')
plt.ylabel('Growth (Base 100)')
plt.grid(True, alpha=0.2)
plt.show()
```
